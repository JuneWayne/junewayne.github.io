# Intro to using APIs in Python

APIs (Application Programming Interfaces) allow different software systems to communicate with each other. In simple terms, think of an API as a waiter in a restaurant. You (the customer) tell the waiter (the API) what you want from the kitchen (the server), and the waiter brings you back your order (the data).

In Python, we can use libraries like `requests` to interact with APIs. The `requests` library simplifies making HTTP requests.

Let's start by installing the `requests` library.

```python
%pip install requests
```

Now that we have the `requests` library installed, let's see how to make a basic request to an API. We'll use a free and open API for testing, like the JSONPlaceholder API, which provides fake data for testing purposes.

A common type of API request is a `GET` request, which is used to retrieve data from a server.

```python
import requests

# The URL of the API endpoint we want to access
api_url = "https://jsonplaceholder.typicode.com/posts/1"

# Make a GET request to the API
response = requests.get(api_url)

# Check if the request was successful (status code 200 means success)
if response.status_code == 200:
    # Get the JSON data from the response
    data = response.json()
    print("Successfully retrieved data:")
    print(data)
else:
    print(f"Error: Unable to fetch data. Status code: {response.status_code}")


# Another useful syntax structure to play with (trial and error)
try:
    response = requests.get(api_url)
    data = response.json()
    print("\nSuccessfully retrieved data:")
    print(data)
except Exception as e:
    print(f"Error: Unable to fetch data. Status code: {response.status_code}")
```

In the code above:
- We import the `requests` library.
- We define the `api_url` which is the specific address of the data we want to get from the API.
- `requests.get(api_url)` sends a GET request to that URL.
- The response from the API is stored in the `response` variable.
- We check the `response.status_code`. A code of 200 is the standard response for successful HTTP requests.
- If the request is successful, we use `response.json()` to parse the JSON data returned by the API into a Python dictionary.
- Finally, we print the retrieved data.

This is just a basic example, but it shows the fundamental steps of making an API request using the `requests` library. Many APIs require authentication or accept different parameters in the request, which are more advanced topics you can explore as you continue learning!

### Beyond GET: Using POST, PUT, and DELETE

While `GET` is used for retrieving data, APIs often support other methods for different operations:

*   **`POST`**: Used to send new data to an API endpoint (e.g., creating a new resource).
*   **`PUT`**: Used to update existing data at an API endpoint.
*   **`DELETE`**: Used to remove data from an API endpoint.

Let's see examples of these using the JSONPlaceholder API.

#### Making a POST Request (Creating Data)

We'll create a new post using the `/posts` endpoint.

We'll also be using `httpbin.org/post` to see the data sent in a POST request.

```python
httpbin_post_url = "https://httpbin.org/post"

# The data we want to send in the request body
data_to_send = {
    "name": "Ethan",
    "job": "Audio Visual Assistant at Newcomb",
    "job responsibility": "put up mics, lecterns, and tape power cables in newcomb ball room"
}

# Make a POST request with the data
response = requests.post(httpbin_post_url, json=data_to_send)

if response.status_code == 200:
    # Get the JSON response from httpbin
    response_data = response.json()
    print("Successfully sent POST request to httpbin.org:")
    # httpbin.org returns the sent data under the 'json' key if sent with json=
    print("Received data:", response_data.get('json'))
else:
    print(f"Error: Unable to send POST request. Status code: {response.status_code}")
```

As you can see in the output, `httpbin.org` received the JSON data we sent in the `POST` request and returned it under the `'json'` key in its response. This confirms that the `requests.post()` method correctly sent your data.

#### Making a PUT request

Using `httpbin.org/put` to see the data sent in a PUT request.

```python
httpbin_put_url = "https://httpbin.org/put"

# The data we want to send in the request body for update
updated_data = {
    "status": "Active",
    "progress": "80%",
    "comment": "Not doing so well"
}

# Make a PUT request with the data
response = requests.put(httpbin_put_url, json=updated_data)
if response.status_code == 200:
    # Get the JSON response from httpbin
    response_data = response.json()
    print("\nSuccessfully sent PUT request to httpbin.org:")
    # httpbin.org returns the sent data under the 'json' key if sent with json=
    print("Received data:", response_data.get('json'))
else:
    print(f"\nError: Unable to send PUT request. Status code: {response.status_code}")
```

Similar to the POST example, `httpbin.org` reflects the data sent in the `PUT` request, confirming that `requests.put()` is working as expected.

#### Making a DELETE request

Using `httpbin.org/delete` to see the details of a DELETE request.

```python
httpbin_delete_url = "https://httpbin.org/delete"

# Make a DELETE request
response = requests.delete(httpbin_delete_url)
if response.status_code == 200:
    # Get the JSON response from httpbin
    response_data = response.json()
    print("\nSuccessfully sent DELETE request to httpbin.org:")
    # httpbin.org returns details about the received request
    print("Received request details:", response_data)
else:
    print(f"\nError: Unable to send DELETE request. Status code: {response.status_code}")
```

For `DELETE`, `httpbin.org` confirms that the DELETE request was received. While you might not send a body with a DELETE request usually, `httpbin.org` shows other details of the request it received.

These examples using `httpbin.org` verify that your Python code using `requests.post()`, `requests.put()`, and `requests.delete()` is correctly formulating and sending these types of HTTP requests with the data you provide. Remember, for actual data persistence and retrieval, you would need to use a real API that supports these operations on its backend.

## So how do I exactly handle return types from APIs?

When you make an API request using `requests`, the response you get back can be in various formats, such as JSON, plain text, HTML, images, or other binary data. The `requests` library provides different ways to access the response body depending on its format. Two common ways are using `.json()` and `.content`.

### Using `.json()`

The `.json()` method is used when the API response body contains data in **JSON format**. This method automatically parses the JSON data and returns it as a Python dictionary or list. This is the most common method when dealing with structured data from APIs.

### Using `.content`

The `.content` attribute is used to access the raw response body as **bytes**. This is useful for non-text responses like images, audio files, or when you need to handle binary data.

### Using `.text`

The `.text` attribute provides the response body as a **string**. Requests tries to guess the encoding based on the response headers or content. This is useful for HTML, XML, or plain text responses.

### How to know which method to use?

The best way to determine how to handle an API response is to check the **`Content-Type` header** in the response. This header tells you the media type of the resource sent in the response body.

You can access the headers using `response.headers`. The `Content-Type` is usually a key in this dictionary.

```python
import requests

# Example using a real API that returns JSON
json_api_url = "https://jsonplaceholder.typicode.com/posts/1"
response_json = requests.get(json_api_url)

print(f"Content-Type for JSON API ({json_api_url}):")
print(response_json.headers.get('Content-Type'))

print("-" * 30)

# Example using an API that returns an image (like the Dog CEO API)
image_api_url = "https://images.dog.ceo/breeds/dingo/n02115641_14117.jpg" # Direct image URL
response_image = requests.get(image_api_url)

print(f"Content-Type for Image URL ({image_api_url}):")
print(response_image.headers.get('Content-Type'))

print("-" * 30)

# Example using an API that might return HTML or plain text
# (Using httpbin.org/html which returns HTML)
html_api_url = "https://httpbin.org/html"
response_html = requests.get(html_api_url)

print(f"Content-Type for HTML API ({html_api_url}):")
print(response_html.headers.get('Content-Type'))

print("-" * 30)
```

Based on the `Content-Type` header:

*   If `Content-Type` is `application/json`, use `.json()`.
*   If `Content-Type` is an image type (e.g., `image/jpeg`, `image/png`, `image/gif`), use `.content`.
*   If `Content-Type` is `text/html`, `text/plain`, or `application/xml`, use `.text`.

Sometimes, APIs might return JSON with a slightly different `Content-Type` (like `application/vnd.api+json`), but `.json()` will often still work. If `.json()` fails with a `JSONDecodeError`, it's a strong indication that the response was not valid JSON, and you might need to inspect `.text` or `.content` and the `Content-Type` header to understand the actual format.

Now, let's create a simple deterministic chatbot workflow with pre-determined prompts and an output

Let's say you are trying to create a virtual tourism chatbot that will answer questions regarding a particular location or tourist sighting (Let's say a UVA tour guide). Here's what your end product should look like:

- A chatbot that is user-friendly
- AI driven
- Produce well-informed answers

```python
# Define the model you want to use
from openai import OpenAI
from google.colab import userdata # Import userdata to get the API key

model = "gpt-4o-mini"

# Define the "Client" and pass the API key
client = OpenAI(api_key=userdata.get('OPENAI_API_KEY'))


# Create a set of pre-determined customer questions
travel_questions = ['Tell me about UVA',
                    'is Charlottesville worth going?',
                    "I honestly don't know much about the difference between UVA and McIntire",
                    "how far away is the Rotunda building from Newcomb hall",
                    'what are some must-see locations around UVA?',
                    'what is UVA best known for?']
# Create a conversation
conversation = [
    {
        # Here we are defining the personality, responsibility, and other role-playing mechanism of the chatbot
        'role': 'system',
        'content': """You are Thomas, you are a friendly virtual tour guide at the University of Virginia,
        delivering valuable insights into UVA's iconic landmarks, school information, and hidden gems, you
        will answer each question concisely and in a story-telling tone."""

    },
    {
        # Here we are providing an example Question-Answer pair for the chatbot to reference
        # This would be the first prompt that initiates the conversation, it is coming from the user's side
        'role':'user',
        'content': 'What does UVA stand for?',
    },
    {
        # This would be the ideal response that comes out of a chatbot
        'role': 'assistant',
        'content': 'Hey there! UVA stands for the University of Virginia, the best public school that exist on the east coast!'
    }
]

# Add each customer question into the conversation as a separate message
for question in travel_questions:
    print(f"\n--- User: {question} ---") # Print the user's question
    conversation.append({'role':'user', 'content':question})
    request = client.chat.completions.create(
        # Input the model that we defined earlier
        model = model,

        # Define which conversation we are using
        messages = conversation,

        # Define the maximum amount of tokens we allow each response to generate
        # Tokens are calculated as: input + output
        max_tokens = 100,

        # Define how creative or 'indeterministic' the chatbot should be
        # 0 = strictly factual
        # 1 = extremely creative
        temperature = 0.8,
    )


    # Capture the chatbot's response
    response = request.choices[0].message.content
    print(f"Bot: {response}") # Print the bot's response

    # Access and print token usage information
    if request.usage:
        print(f"Tokens Used for this turn:")
        print(f"  Prompt Tokens: {request.usage.prompt_tokens}")
        print(f"  Completion Tokens: {request.usage.completion_tokens}")
        print(f"  Total Tokens: {request.usage.total_tokens}")

    # We append the earlier conversation history back into the chatbot to maintain context
    # This also enables follow-ups, as the chatbot can refer to what was said earlier on in the history
    response_dict = {'role':'assistant', 'content':response}
    conversation.append(response_dict)

print("\n--- Conversation Ended ---")
```

In the code block above, we created a simple, deterministic chatbot conversation flow. Here's how it works:

1.  **Define the Model and Client:** We specified the OpenAI model (`gpt-4o-mini`) and created a client object to interact with the OpenAI API.
2.  **Create Pre-determined Questions:** We defined a list `travel_questions` containing the specific questions we want the chatbot to "answer" in this demonstration.
3.  **Initialize the Conversation:** We started the `conversation` list with a system message (`'role': 'system'`) to set the chatbot's persona and instructions. We also included an example user-assistant turn (`'role': 'user'` and `'role': 'assistant'`) to provide the model with a format and style to follow (this is often called few-shot prompting).
4.  **Loop Through Questions and Get Responses:** We iterated through the `travel_questions` list. For each question:
    *   We appended the user's question (`{'role':'user', 'content':question}`) to the `conversation` history.
    *   We made a call to the OpenAI API (`client.chat.completions.create`) with the current `conversation` history. The model generated a response based on its instructions and the conversation so far.
    *   We extracted the generated response text.
    *   We printed the response.
    *   **Crucially**, we appended the assistant's response (`{'role':'assistant', 'content':response}`) back to the `conversation` list. This ensures that the history is maintained for the next turn in the loop, allowing the model to build on the previous responses.

This setup creates a scripted interaction where the chatbot responds to a fixed sequence of questions, demonstrating how conversation history is passed to the API to maintain context.

Now, let's create a simple loop that allows you to chat with the OpenAI model in the terminal. We'll use the `gpt-3.5-turbo` model, which is suitable for chat.

```python
import sys

# Ensure the API key is set and create the client instance
try:
    api_key = userdata.get('OPENAI_API_KEY')
    if not api_key:
        raise ValueError("OPENAI_API_KEY not found in Colab Secrets.")
    # Create the OpenAI client instance
    client = OpenAI(api_key=api_key)
except Exception as e:
    print(f"Error setting up OpenAI client: {e}")
    print("Please make sure you have added your OPENAI_API_KEY to Colab Secrets.")
    sys.exit()


print("Welcome to our simple generalist chatbot. Type 'quit' or 'exit' to end the conversation.")

# Get the model name from user input
model_name = input("Enter the OpenAI model name (e.g., gpt-3.5-turbo, gpt-4): ")

messages = [{"role": "system", "content": "You are a helpful assistant."}]

while True:
    user_input = input("You: ")
    if user_input.lower() in ['quit', 'exit']:
        break

    messages.append({"role": "user", "content": user_input})

    try:
        # Use the client instance to create chat completions
        response = client.chat.completions.create(
            model=model_name,
            messages=messages,
            max_tokens=150,
            temperature=0.8
        )
        assistant_response = response.choices[0].message.content
        print(f"Bot: {assistant_response}")
        messages.append({"role": "assistant", "content": assistant_response})

    except Exception as e:
        print(f"An error occurred: {e}")

print("Chat session ended.")
```

In this chat loop:
- We ensure the API key is set and create an instance of the `OpenAI` client, storing it in the `client` variable. This is the recommended way to interact with the OpenAI API.
- We initialize a list called `messages` to keep track of the conversation history. The first message sets the role of the system.
- The code prompts you to enter the OpenAI model name you want to use.
- The `while True` loop continues until you type 'quit' or 'exit'.
- Inside the loop, it prompts you for input (`input("You: ")`).
- Your input is added to the `messages` list with the role "user".
- `client.chat.completions.create()` sends the conversation history to the OpenAI API using the specified model.
- The model's response is extracted and printed with the prefix "Bot: ".
- The bot's response is also added to the `messages` list with the role "assistant" to maintain context for future turns.
- Error handling is included to catch potential issues with the API call.

This provides a basic command-line interface for interacting with the OpenAI model as a chatbot, using the modern client instance approach. You can expand on this by adding more complex system messages, handling different user intents, or integrating it into a more sophisticated application.

# Multi-Agent Implementation

### What is a Multi-Agent System?

In the context of AI and large language models (LLMs), a multi-agent system is a collection of individual AI agents that work together to achieve a common goal or solve a complex problem. Instead of a single AI trying to do everything, a multi-agent system breaks down the problem into smaller parts and assigns each part to a specialized agent with a specific role, expertise, and sometimes, access to particular tools.

These agents can interact with each other, delegate tasks, share information, and collaborate in a structured way. This approach can be particularly effective for tasks that require different skills or perspectives, mimicking how a team of humans might collaborate on a project. CrewAI is a framework that helps you build and manage such multi-agent systems.

### Defining Agents in CrewAI

In CrewAI, an agent is an autonomous unit with a specific role, goal, and capabilities (tools). You can define agents with different responsibilities to work together on a task.

Let's say we want to create a digital market research team:
we'll need a

```python
from crewai import Agent, LLM
from google.colab import userdata # Import userdata to get the API key

# Get the market field from user input
market_field = input("Enter the market field you want the team to research (e.g., renewable energy, artificial intelligence): ")

# Define the LLM to be used by the agents
# Using gpt-4o-mini as an example, fetching API key from Colab Secrets
llm = LLM(
    model="gpt-4o-mini",
    temperature=0.7,
    api_key=userdata.get('OPENAI_API_KEY') # Fetch API key from Colab Secrets
)

# Define the Senior Market Researcher agent
senior_researcher = Agent(
  role='Senior Market Researcher',
  goal=f'Analyze market trends and identify key opportunities in the {market_field} sector',
  backstory=f"""You are a Senior Market Researcher with extensive experience in
  analyzing market data, identifying emerging trends, and uncovering key insights within the {market_field} sector.
  You are skilled in data collection, analysis, and presenting findings clearly, you have to present your final analysis
  in an extremely concise manner.""",
  verbose=True,
  allow_delegation=False,
  llm=llm
)
```

Now that we have our agents, we need to define the tasks they will perform. Tasks are specific actions or objectives assigned to an agent.

```python
from crewai import Task

# Define the Market Trend Analysis task for the Senior Researcher
market_research_task = Task(
  description=f"""Conduct a comprehensive analysis of the current market trends
  in the {market_field} sector. Identify key drivers, growth areas, and
  potential challenges. Focus on recent developments.
  Your final output should be a detailed analysis report on current trends in the {market_field} sector.""",
  expected_output=f'A detailed market analysis report on current trends in the {market_field} sector.',
  agent=senior_researcher
)

# Define the Data Processing and Visualization Support task for the Data Analyst
data_analysis_task = Task(
  description=f"""Based on the market analysis report from the Senior Market Researcher on the {market_field} sector,
  identify key data points and insights that can be presented visually.
  Prepare a summary of the key findings and suggest potential data visualizations
  (e.g., charts, graphs) to support the report.
  Compile your final answer as a summarized list of key data points and visualization suggestions for the {market_field} market.""",
  expected_output=f'A summarized list of key data points and suggestions for data visualizations based on the research report for the {market_field} sector.',
  agent=data_analyst
)

# Define the Competitive Analysis task for the Competitive Analyst
competitive_analysis_task = Task(
  description=f"""Analyze the key competitors operating within the {market_field} sector.
  Identify their market share, main products/services, recent strategies, and strengths and weaknesses.
  Use the market trends analysis report (from the Senior Market Researcher) and any relevant data insights (from the Market Data Analyst) to inform your analysis.
  Your final output should be a report on the competitive landscape in the {market_field} sector.""",
  expected_output=f'A report detailing the competitive landscape, key players, and their strategies in the {market_field} sector.',
  agent=competitive_analyst
)

# Define the Compile Final Report task for the Report Writer
report_writing_task = Task(
  description=f"""Take the market analysis report from the Senior Market Researcher, the
  key findings/visualization suggestions from the Market Data Analyst, and the competitive analysis report from the Competitive Analyst regarding the {market_field} sector.
  Compile a comprehensive and well-structured market research report.
  Ensure the report is clear, concise, and presents the key insights effectively for stakeholders interested in the {market_field} market.
  Include the summarized key findings, reference potential visualizations, and integrate the competitive landscape analysis.
  Compile your final answer in a markdown file.""",
  expected_output=f'A final, compiled market research report in markdown format on the {market_field} sector, incorporating findings from the Senior Market Researcher, Market Data Analyst, and Competitive Analyst.',
  agent=report_writer
)
```

### Forming a Crew

In CrewAI, a crew is a group of agents working together on a set of tasks. We define the crew and assign the tasks to them. The framework then orchestrates the execution of these tasks by the agents.

```python
from crewai import Crew, Process

# Instantiate a Crew for the market research team with the agents that now have llm configured
market_research_crew = Crew(
  agents=[senior_researcher, data_analyst, competitive_analyst, report_writer],
  tasks=[market_research_task, data_analysis_task, competitive_analysis_task, report_writing_task],
  process=Process.sequential  # Tasks are executed in a sequence
)
```

### Running the Crew

Now that the crew is assembled with their tasks, we can kick off the process. The `kickoff()` method starts the execution of the tasks by the agents.

```python
# Kickoff the process for the market research crew
import os
from google.colab import userdata

os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')

result = market_research_crew.kickoff()

print("## Here is the final result:")
print(result)
```

### Explanation of the Code

In this multi-agent example:
- We defined three agents: a `senior_researcher`, a `data_analyst`, and a `report_writer`, each with a specific role, goal, and backstory.
- We defined three tasks: `market_research_task`, `data_analysis_task`, and `report_writing_task`, each assigned to a specific agent and with a clear description and expected output.
- We created a `Crew` with the defined agents and tasks, specifying the process as `sequential`, meaning the tasks will be executed one after the other in the order they are listed.
- Finally, we called `market_research_crew.kickoff()` to start the multi-agent process. CrewAI orchestrates the agents to perform their tasks, passing the output of one task as input to the next, simulating a workflow.

This demonstrates a basic multi-agent system using CrewAI. You can create more complex workflows with different types of agents, tasks, and processes (like parallel execution) to tackle more sophisticated problems.
